```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print("Libraries loaded")
```

```python
# ── Dataset 1: Patient Admissions ──────────────────────────────
df_admissions = pd.read_csv('../data/raw/healthcare_dataset.csv')

print("=== SHAPE ===")
print(df_admissions.shape)

print("\n=== COLUMNS & TYPES ===")
print(df_admissions.dtypes)

print("\n=== FIRST 3 ROWS ===")
df_admissions.head(3)
```

```python
print("=== NULL VALUES ===")
print(df_admissions.isnull().sum())

print("\n=== UNIQUE VALUE COUNTS ===")
for col in df_admissions.columns:
    print(f"{col}: {df_admissions[col].nunique()} unique values")
```

### Data Quality Check Summary

The dataset contains 55,500 patient admission records and 15 columns.  
The null-value check shows that there are no missing values in any column, so no imputation or row removal is required at this stage.

The unique-value count helps identify the nature of each column. Columns such as Gender, Blood Type, Medical Condition, Insurance Provider, Admission Type, Medication, and Test Results have a small number of unique values, which makes them useful categorical fields for grouping and dashboard analysis.

Columns such as Name, Doctor, and Hospital have a high number of unique values. The Billing Amount column  has many unique values, which is expected because it is a continuous numeric field.

```python
print("=== MEDICAL CONDITIONS ===")
print(df_admissions['Medical Condition'].value_counts())

print("\n=== ADMISSION TYPES ===")
print(df_admissions['Admission Type'].value_counts())

print("\n=== TEST RESULTS ===")
print(df_admissions['Test Results'].value_counts())

print("\n=== INSURANCE PROVIDERS ===")
print(df_admissions['Insurance Provider'].value_counts())
```

### Categorical Distribution Summary

The categorical distribution check shows that the main healthcare fields are generally balanced. Medical conditions are distributed almost evenly across Arthritis, Diabetes, Hypertension, Obesity, Cancer, and Asthma. This balance makes the dataset useful for comparing admission patterns and billing behaviour across different condition groups.

Admission types are also evenly distributed across Elective, Urgent, and Emergency visits. This allows fair comparison between planned admissions and more urgent hospital visits.

Test results are split almost equally across Abnormal, Normal, and Inconclusive outcomes. This can support later analysis of test-result trends by condition, admission type, medication, or insurance provider.

Insurance providers are also relatively balanced, with Cigna, Medicare, UnitedHealthcare, Blue Cross, and Aetna all having similar record counts. 

```python
# ── Dataset 2: ER Wait Times ──────────────────────────────
df_er = pd.read_csv('../data/raw/ER Wait Time Dataset.csv')

print("=== SHAPE ===")
print(df_er.shape)

print("\n=== COLUMNS & TYPES ===")
print(df_er.dtypes)

print("\n=== NULL VALUES ===")
print(df_er.isnull().sum())

print("\n=== URGENCY LEVELS ===")
print(df_er['Urgency Level'].value_counts())

print("\n=== REGIONS ===")
print(df_er['Region'].value_counts())

print("\n=== PATIENT OUTCOMES ===")
print(df_er['Patient Outcome'].value_counts())

print("\n=== WAIT TIME STATS (minutes) ===")
print(df_er['Total Wait Time (min)'].describe())
```

### ER Wait-Time Dataset Summary

The ER wait-time dataset contains 5,000 emergency room visit records and 19 columns. The initial quality check shows that there are no missing values across any of the columns.

The urgency levels are well balanced across Low, Medium, High, and Critical cases, which allows fair comparison of wait-time patterns by urgency. The dataset contains more Urban visits than Rural visits, making region an important variable for analyzing differences in healthcare access and patient flow.

Most patients were discharged, followed by admitted patients. A smaller group left without being seen, which is an important operational outcome because it may indicate long wait times, overcrowding, or access challenges.

The average total wait time is approximately 82 minutes, while the median wait time is 60 minutes. Since the average is higher than the median, the wait-time distribution appears right-skewed, meaning that some very long waits are increasing the overall average.

```python
# ── Dataset 3: CIHI Provincial Wait Times ──────────────────────────────
df_cihi = pd.read_excel('../data/raw/wait-times-priority-procedures-in-canada-2025-data-tables-en.xlsx',
                        sheet_name='Table 1',
                        header=1)

# Drop the empty trailing columns
df_cihi = df_cihi.dropna(axis=1, how='all')

print("=== SHAPE ===")
print(df_cihi.shape)

print("\n=== COLUMNS ===")
print(df_cihi.columns.tolist())

print("\n=== PROVINCES ===")
print(df_cihi['Province'].unique())

print("\n=== PROCEDURES ===")
print(df_cihi['Indicator'].unique())

print("\n=== METRICS ===")
print(df_cihi['Metric'].unique())

print("\n=== YEAR RANGE ===")
print(sorted(df_cihi['Data year'].dropna().astype(str).unique()))
print("\n=== SAMPLE ROWS ===")
df_cihi.head(5)
```

### CIHI Provincial Wait Times Dataset Summary

The CIHI wait-times dataset contains 18,965 rows and includes provincial, regional, and Canada-level wait-time results for priority procedures. The dataset includes procedures such as cancer surgeries, cataract surgery, CT scans, MRI scans, hip replacement, knee replacement, and radiation therapy.

The `Metric` column is especially important because the meaning of `Indicator result` changes depending on the selected metric. For example, when the metric is `50th Percentile` or `90th Percentile`, the result represents wait time in days. When the metric is `Volume`, the result represents the number of cases. When the metric is `% Meeting Benchmark`, the result represents benchmark performance.

Some extra `Unnamed` columns appeared because the source is a formatted Excel workbook. These columns should be removed before further analysis. The dataset also contains some missing province, indicator, and metric values, which are likely blank or note rows from the Excel file and should be cleaned before analysis.
